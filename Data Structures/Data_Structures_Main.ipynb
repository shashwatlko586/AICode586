# **Count Vowels and Consonants in a String**

# **Instructions**

Write a program that takes a string as input and counts the number of vowels and consonants in it.

Vowels include
 'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'.
Consonants are all other alphabetic letters (i.e., all letters except vowels).
All non-alphabetic characters (e.g., spaces, digits, punctuation) in the input string should be disregarded.


Input Format:

A string 's' consists of any valid ASCII character.


Output Format:

Count of vowels in the first line.
Count of consonants in the second line.


Constraints:

1 ≤ len(s) ≤ 105
The string may contain uppercase and lowercase characters, spaces, and punctuation, but only alphabetic characters should be counted.




Sample Input 1:

hello



# **Sample Output 1:**

2

3



Sample Input 2:

Hi,Upgrad...!


Sample Output 2:

3

5

```python
# Taking input
s = input()

def count_vowels_and_consonants(s):
    # Write your code here
    s = s.lower()
    count_vowels = 0
    count_consonants = 0
    vowels = 'aeiou'
    for char in s:
        if char.isalpha() :
            if char in vowels :
                count_vowels += 1
            else :
                count_consonants += 1
    return count_vowels , count_consonants

# Print the Output
vowels, consonants = count_vowels_and_consonants(s)
print(vowels)
print(consonants)
```

Input Format:

The program accepts three sorted lists as input, one list per line.


Output Format:

If common elements exist, output them as space-separated integers in a single line.
If no common elements exist, output "None" (without quotes).

Each list is sorted in non-decreasing order.

Lists may contain duplicates
No Nested loops




Sample Input:

[1, 5, 10, 20, 40, 80]

[6, 7, 20, 80, 100]  

[3, 4, 15, 20, 30, 70, 80, 120]



# **Sample Output:**

```python
# Import the literal_eval library to evaluate string input as a Python safely
from ast import literal_eval

# Taking input
list1 = literal_eval(input())
list2 = literal_eval(input())
list3 = literal_eval(input())

# Efficient linear traversal using three pointers
def find_common_elements(l1, l2, l3):
    # Write your code here
    l4 = l1 + l2+ l3
    l5 = []
    n = len(l4)
    for i in range(n) :
        if l4.count(l4[i]) >= 3 and l1.count(l4[i]) >= 1 and l2.count(l4[i]) >= 1 and l4.count(l4[i]) >= 1 :
            if l4[i] not in l5 :
                l5.append(l4[i])
                #print(l5)
    return l5

common_elements = find_common_elements(list1, list2, list3)

# Print the output
print(" ".join(map(str, common_elements)) if common_elements else "None")

```

# **Find Most Frequent Vowel and Consonants**

# **Instructions**

Write a program that takes a string 's' consisting of only lowercase English letters (from 'a' to 'z') and performs the following operations:

Identify the vowel (one of 'a', 'e', 'i', 'o', or 'u')  that appears with the highest frequency in the string.
Identify the consonant (i.e., all other lowercase letters except vowels) that appears with the highest frequency.
Compute and return the sum of the frequencies of the most frequent vowel and the most frequent consonant.


Note: If multiple vowels or consonants have the same maximum frequency, you may choose any one of them.



The frequency of a letter 'x' is the number of times it occurs in the string.



Input Format:

A string 's' consisting of lowercase English letters only.



Output Format:

Return the sum of the frequencies of the most frequent vowel and consonant.



Constraints:

1 <= len(s) <= 100
's' consists of lowercase English letters only.




Sample Input 1:

"aeiaeia"



# **Sample Output 1:**

3



Sample Input 2:

"successes"



Sample Output 2:

6

```python
# Taking input
s = input()

def max_vowel_consonant_sum(s):
# Write your code here
    s = s.lower()
    max_count_vowels = 0
    max_count_consonants = 0
    current_count_vowels = 0
    current_count_consonants = 0
    vowels = 'aeiou'
    for char in s:
        if char.isalpha() :
            if char in vowels :
                current_count_vowels = s.count(char)
                if current_count_vowels > max_count_vowels :
                    max_count_vowels = current_count_vowels
            else :
                current_count_consonants = s.count(char)
                if current_count_consonants > max_count_consonants :
                    max_count_consonants = current_count_consonants
    return max_count_vowels + max_count_consonants

# Print the output
print(max_vowel_consonant_sum(s))
```

# **Find Longest Word in a Sentence**

# **Instructions**

Write a program that defines a function 'longest_word(sentence)' which returns the longest word in the sentence. If multiple words have the same maximum length, return the first one among them.



Input Format:

A single line containing a sentence with words separated by spaces.



Output Format:

A single word that is the longest in the sentence.



Constraints:

Sentence contains at least one word
Sentence length ≤ 100 characters
All words consist of lowercase or uppercase English alphabets only
The program should preserve the original casing of words. No case conversion is required.


Sample Input 1:

The elephant met a caterpillar



# **Sample Output 1:**

caterpillar



Explanation:

The word "caterpillar" has 11 characters, which is longer than any other word in the sentence. Hence, it is returned.





Sample Input 2:

I love competitive programming



Sample Output 2:

competitive



Explanation:

Both "competitive" and "programming" have 11 characters, the longest in the sentence. However, since "competitive" appears first, it is returned.

```python
# Taking input
sentence = input()

def longest_word(sentence):
    # Write your code here
    words = sentence.split()
    longest = words[0]
    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

# Print the output
print(longest_word(sentence))
```

# **Find All Duplicates in a List of Strings**

# **Instructions**

Write a function 'find_duplicate_strings(input_list)' that accepts a list of strings. The function must identify and return a new list containing only those strings that occur at least twice within the 'input_list'. The returned list must consist of unique strings, ordered by their first appearance as a duplicate in the original 'input_list'.



Input Format:

A list of strings, where each string contains only lowercase or uppercase alphabetical characters.
The list can contain duplicate entries.
Output Format:

A list of strings that appear more than once in the input list.
Each string in the output should appear only once, and in the order of its first duplication.


Constraints:

1 ≤ len(input_list) ≤ 100.
Each string contains only alphabetic characters.
Case-sensitive comparison ("A" ≠ "a")
Return order must match the order in which duplicates appear.




Sample Input 1:

["a", "b", "a", "c", "b"]


# **Sample Output 1:**

['a', 'b']



Sample Input 2:

["python", "Python", "PYTHON", "python"]



Sample Output 2:

['python']

```python
# Import the literal_eval library to evaluate string input as a Python safely
from ast import literal_eval

# Taking input
lst = literal_eval(input())

def find_duplicates(lst):
    # Write your code here
    dup_val = []
    for str_val in lst :
        if lst.count(str_val) > 1 :
            if str_val not in dup_val :
                dup_val.append(str_val)
    return dup_val
# Print the output
print(find_duplicates(lst))
```
