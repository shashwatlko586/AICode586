# **String Compression**

# **Instructions**

Write a program to perform basic string compression using the counts of repeated characters. For each group of consecutive repeating characters in the input string, replace it with the character followed by the number of repetitions. If the compressed string is not shorter than the original string, return the original string.





Input Format:

A single string 's' consisting of only lowercase English letters.



Output Format:

A single string: the compressed version of 's' if it is shorter than 's'; otherwise, 's'.



Constraints:

1 <= len(s) <= 1000

's' contains only lowercase English alphabets [a-z]



Sample Input 1:

aabcccccaaa



# **Sample Output 1:**

a2b1c5a3



Sample Input 2:

abcd



Sample Output 2:

abcd

```python
# Taking Input
s = input()

# Write your code here
result = ''
ch1 = ch2 = ''
count = 0
for ch3 in s :
    if ch3 == ch2 :
        count += 1
        continue
    count_s = 0
    for ch4 in s[ count : len(s)]  :
        if ch4 != ch3 :
            break
        else :
            count_s += 1
    result = result + ch3 + str( count_s )
    ch1 = ch1 + ch3 + ( str( count_s ) if count_s > 1 else '')
    ch2 = ch3
    count += 1
if ch1 == s :
    result = s
# Print the Output
print(result)
```

# **Check if a String is a Palindrome**

# **Instructions**

Write a Python program that checks whether a given string is a palindrome.



A palindrome is a word, phrase, number, or other sequence of characters that reads the same forward and backward.



The program must ignore the following when checking for a palindrome:

Spaces
Punctuation (like commas, periods, exclamation marks)
Case (uppercase vs. lowercase letters)


Input Format:

A string 's' that may contain letters, spaces, and punctuation.



Output Format:

True if the string is a palindrome (ignoring spaces, punctuation, and case).
False otherwise.


Constraints:

The input string contains printable characters (letters, spaces, punctuation).



Sample Input :

"madam"



# **Sample Output:**

True

```python
# Code Pattern 1
# Taking input
s=input() # A man a plan a canal Panama

def is_palindrome(s):
    # Write your code here
    if s.upper().replace(' ', '').replace('!', '').replace(',', '').replace('.', '') == s[  : : -1 ].upper().replace(' ', '').replace('!', '').replace(',', '').replace('.', '') :
        return True
    else :
        return False

# Print the output
print(is_palindrome(s))
```

```python
# Code Pattern 2
# Taking input
s=input()

def is_palindrome(s):
    # Write your code here
    s = s.lower()
    cleaned_s = ""

    for char in s:
        if char.isalnum(): # Keep only alphanumeric characters
            cleaned_s += char
    return cleaned_s == cleaned_s[::-1]

# Print the output
print(is_palindrome(s))
```

# **List Index Search**

# **Instructions**

 Write a program that takes a list of elements and finds the index of a given element in the list. If the element does not exist in the list, the program should return -1.



Input Format:

A list of integers.
An integer to search for in the list.


Output Format:

The index of the element in the list, or -1 if the element is not found.


Constraints:

The list will contain integers.
The search element is an integer.




Sample Input:

[10, 20, 30, 40, 50]  

30



# **Sample Output:**

2

```python
# Import literal_eval library to safely evaluate string input as a Python
from ast import literal_eval

# Taking input
lst = literal_eval(input())
search_element = int(input())

def find_index(lst, search_element):
        # Write your code here
        if search_element not in lst :
                return '-1'
        else :
                return lst.index(search_element)

# Print the output
print(find_index(lst, search_element))
```

# **Anagram Check**

# **Instructions**

Write a function ‘are_anagrams(str1, str2)’ that takes two strings ‘str1’ and ‘str2’ as input and returns True if they are anagrams of each other, and False otherwise. The function should be case-insensitive and should ignore spaces between words.



An anagram is when two strings contain the same characters, but possibly in a different order.




Input Format:

Two strings, ‘str1’ and ‘str2’, which may contain spaces and are case-insensitive.



Output Format:

True if the two strings are anagrams, and False otherwise.



Constraints:

The input strings can be of length between 0 and 100.
The input strings can contain alphabetic characters (uppercase and lowercase) and spaces.




Sample input:

listen

silent



# **Sample Output:**

True

```python
# Code Pattern 1
# Taking input
str1 = input()
str2 = input()

def are_anagrams(str1, str2):
    # Write your code here
    str1 = list(str1.upper().replace(' ', ''))
    str2 = list(str2.upper().replace(' ', ''))
    str1.sort()
    str2.sort()
    if str1 == str2 :
        return True
    else :
        return False

# Print the result
print(are_anagrams(str1, str2))
```

```python
# Code Pattern 2
# Taking input
str1 = input()
str2 = input()

def are_anagrams(str1, str2):
    # Write your code here
    str1 = str1.replace(' ', '').lower()
    str2 = str2.replace(' ', '').lower()
    # Sorting both strings and comparing them
    return sorted(str1) == sorted(str2)

# Print the result
print(are_anagrams(str1, str2))
```
