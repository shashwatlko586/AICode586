# **String Compression**

# **Instructions**

Write a program to perform basic string compression using the counts of repeated characters. For each group of consecutive repeating characters in the input string, replace it with the character followed by the number of repetitions. If the compressed string is not shorter than the original string, return the original string.





Input Format:

A single string 's' consisting of only lowercase English letters.



Output Format:

A single string: the compressed version of 's' if it is shorter than 's'; otherwise, 's'.



Constraints:

1 <= len(s) <= 1000

's' contains only lowercase English alphabets [a-z]



Sample Input 1:

aabcccccaaa



# **Sample Output 1:**

a2b1c5a3



Sample Input 2:

abcd



Sample Output 2:

abcd

```python
# Taking Input
s = input()

# Write your code here
result = ''
ch1 = ch2 = ''
count = 0
for ch3 in s :
    if ch3 == ch2 :
        count += 1
        continue
    count_s = 0
    for ch4 in s[ count : len(s)]  :
        if ch4 != ch3 :
            break
        else :
            count_s += 1
    result = result + ch3 + str( count_s )
    ch1 = ch1 + ch3 + ( str( count_s ) if count_s > 1 else '')
    ch2 = ch3
    count += 1
if ch1 == s :
    result = s
# Print the Output
print(result)
```

# **Check if a String is a Palindrome**

# **Instructions**

Write a Python program that checks whether a given string is a palindrome.



A palindrome is a word, phrase, number, or other sequence of characters that reads the same forward and backward.



The program must ignore the following when checking for a palindrome:

Spaces
Punctuation (like commas, periods, exclamation marks)
Case (uppercase vs. lowercase letters)


Input Format:

A string 's' that may contain letters, spaces, and punctuation.



Output Format:

True if the string is a palindrome (ignoring spaces, punctuation, and case).
False otherwise.


Constraints:

The input string contains printable characters (letters, spaces, punctuation).



Sample Input :

"madam"



# **Sample Output:**

True

```python
# Code Pattern 1
# Taking input
s=input() # A man a plan a canal Panama

def is_palindrome(s):
    # Write your code here
    if s.upper().replace(' ', '').replace('!', '').replace(',', '').replace('.', '') == s[  : : -1 ].upper().replace(' ', '').replace('!', '').replace(',', '').replace('.', '') :
        return True
    else :
        return False

# Print the output
print(is_palindrome(s))
```

```python
# Code Pattern 2
# Taking input
s=input()

def is_palindrome(s):
    # Write your code here
    s = s.lower()
    cleaned_s = ""

    for char in s:
        if char.isalnum(): # Keep only alphanumeric characters
            cleaned_s += char
    return cleaned_s == cleaned_s[::-1]

# Print the output
print(is_palindrome(s))
```

# **Anagram Check**

# **Instructions**

Write a function ‘are_anagrams(str1, str2)’ that takes two strings ‘str1’ and ‘str2’ as input and returns True if they are anagrams of each other, and False otherwise. The function should be case-insensitive and should ignore spaces between words.



An anagram is when two strings contain the same characters, but possibly in a different order.




Input Format:

Two strings, ‘str1’ and ‘str2’, which may contain spaces and are case-insensitive.



Output Format:

True if the two strings are anagrams, and False otherwise.



Constraints:

The input strings can be of length between 0 and 100.
The input strings can contain alphabetic characters (uppercase and lowercase) and spaces.




Sample input:

listen

silent



# **Sample Output:**

True

```python
# Code Pattern 1
# Taking input
str1 = input()
str2 = input()

def are_anagrams(str1, str2):
    # Write your code here
    str1 = list(str1.upper().replace(' ', ''))
    str2 = list(str2.upper().replace(' ', ''))
    str1.sort()
    str2.sort()
    if str1 == str2 :
        return True
    else :
        return False

# Print the result
print(are_anagrams(str1, str2))
```

# **Biggest Product from Two Numbers**

# **Instructions**

Write a function ‘max_product(lst)’ that takes a list ‘lst’ of integers and returns the maximum product of any two distinct elements from the list. The list can contain both positive and negative integers.



Input Format:

A list of integers.



Output Format:

The maximum product of any two distinct elements from the list.



Constraints:

The list will contain at least two integers.
The integers in the list may be positive, negative, or zero.


Sample Input:

[1, 10, -5, 1]



# **Sample Output:**

10

```python
# Import literal_eval library to safely evaluate string input as a Python
from ast import literal_eval

# Taking input
lst = literal_eval(input())

def max_product(lst):
    # Write your code here
    lst.sort()
    return lst[-1] * lst[-2] if lst[-1] * lst[-2] > lst[0] * lst[1] else lst[0] * lst[1]

# Print the output
print(max_product(lst))
```

# **Return the Median value**

# **Instructions**

You are given two lists of integers; 'list1' and 'list2'. Merge these lists into one sorted list and return the median value.



If the total number of elements is odd, return the middle element.
If even, return the average of the two middle elements.


This task requires handling both odd and even-length combined lists.



Input Format:

Two lines:

A list of integers (list1)
Another list of integers (list2)


Output Format:

A single float or integer representing the median



Constraints:

1 ≤ len(list1) + len(list2) ≤ 1000
Elements in the list can be positive, negative, or zero


Sample Input:

[1, 3, 5, 7, 9]

[2, 4, 6, 8, 10]



# **Sample Output:**

5.5

```python
# Import literal_eval library to safely evaluate string input as a Python
from ast import literal_eval

# Taking input
list1 = literal_eval(input())
list2 = literal_eval(input())

def find_median_lists(list1, list2):
    # Write your code here
    combined = sorted(list1 + list2)
    n = len(combined)
    # median value
    return combined[n // 2] if n % 2 != 0 else ( ( combined[n // 2 - 1] ) + ( combined[n // 2] )) / 2

# Print the output
print(find_median_lists(list1, list2))
```

Input Format:

Output Format:

A single integer that represents the count of elements strictly greater than the average of the list.



Constraints:

The list contains at least 1 and at most 100 integers.
Each integer lies between -10⁵ and 10⁵.


Sample Input 1:

[1, 2, 3, 4, 5]



Sample Output 1:

2



Explanation:

Sample Input 2:

[10, 10, 10, 10]

```python
# Import literal_eval library to safely evaluate string input as a Python
from ast import literal_eval

# Taking input
numbers = literal_eval(input())

def count_above_average(numbers):
    # Write your code here
    return sum(1 for x in numbers if x > ( sum(numbers) / len(numbers)  ) )

# Print the output
print(count_above_average(numbers))
```

```python
# Code Pattern 2
def remove_duplicates(lst):
    # Handle the empty list edge case
    if not lst:
        return 0

    # 'i' is the index of the last known unique element
    i = 0

    # 'j' moves through the list to find new unique elements
    for j in range(1, len(lst)):
        # If we find a new value that is different from our last unique one...
        if lst[j] != lst[i]:
            i += 1          # Move the unique pointer forward
            lst[i] = lst[j] # Overwrite the duplicate with the new unique value

    # The length of the unique part is the index i + 1
    # We slice the list to actually remove the leftover elements at the end
    del lst[i + 1:]

    return len(lst)

# Sample Execution
sample_list = [1, 1, 1, 1, 2, 2, 2, 2, 3]
new_length = remove_duplicates(sample_list)

print(f"New Length: {new_length}") # Output: 3
print(f"Modified List: {sample_list}") # Output: [1, 2, 3]
```

# **Check for Balanced Parentheses**

# **Instructions**

Write a function ‘is_balanced(s)’ that checks if the parentheses in the given string are balanced. The function should also check for balanced brackets [] and curly braces {} in addition to the regular parentheses ().



A string is considered balanced if:

Every opening bracket has a corresponding closing bracket.
The brackets are closed in the correct order.


Input Format:

A single string ‘s’ consisting of characters such as (, ), {, }, [, ], and other non-bracket characters.



Output Format:

True if the string has balanced brackets and False otherwise.



Constraints:

 1<=len(s)<=105
The string can contain any combination of the characters (), {}, [], and other non-bracket characters.




Sample Input:

{[()]}



# **Sample Output:**

True

```python
# Taking input
s = input()

def is_balanced(s):
    # Write your code here
    brackets = [char for char in s if char in "()[]{}"]
    n = len(brackets)
    if n % 2 != 0:
        return False
    for i in range(n // 2):
        left_char = brackets[i]
        right_char = brackets[n - 1 - i]

        # Check if the right character is the partner of the left
        if ( left_char == "(" and right_char == ")" ) or ( left_char == "{" and right_char == "}" ) or ( left_char == "[" and right_char == "]" ) :
            return True
        else : return False

# Print the output
print(is_balanced(s))
```

# **Flatten a List of Lists**

# **Instructions**

Write a program that flattens a nested list into a single list using List Comprehension. The program should take the nested list as input and return the flattened version of the list.



Input Format:

A nested list.



Output Format:

A flattened list that contains all the elements from the sublists of lists in the same order.



Constraints:

The input is a nested list (list of lists).
Both the outer list and its sublists can be empty.
Sublist elements can be of any data type (e.g., Integers, Strings, Floats, Booleans).




Sample Input:

[[1, 2, 3], [4, 5], [6]]



# **Sample Output:**

[1, 2, 3, 4, 5, 6]

```python
# Import the literal_eval library to safely evaluate string input as a Python
from ast import literal_eval

# Taking input
lists = literal_eval(input())

def flatten_lists(lists):
    # Write your code here
    flattened_list = []
    for ele in lists:
        if isinstance(ele, list):
            flattened_list.extend(flatten_lists(ele))
        else:
            flattened_list.append(ele)

    return flattened_list

# Print the output
print(flatten_lists(lists))
```

# **Instructions**

Given two strings s and t, write a program to determine if they are isomorphic.



Two strings 's' and 't' are isomorphic if the characters in 's' can be replaced to get 't'.



All occurrences of a character must be replaced with another character while preserving the order of characters. No two characters may map to the same character, but a character may map to itself.



Input Format:

The program expects two strings as input.



Output Format:

True if two strings are isomorphic, False otherwise.



Constraints:

1 <= len(s)<= 5 * 104
len(t) == len(s)
's' and 't' consist of any valid ASCII character.




Sample Input 1:

"egg "

"add"



# **Sample Output 1:**

True



Sample Input 2:

"foo"

"bar"



Sample Output 2:

False

```python
# Taking input
s = input()
t = input()

def are_isomorphic(s, t):
    # Write your code here
    s_lst = list(s.replace('"', '').strip())
    t_lst = list(t.replace('"', '').strip())
    #print(len(s_lst))
    #print(len(t_lst))
    if len(s_lst) == 0 or len(t_lst) == 0 :
        #print("in 0")
        return True
    elif len(s_lst) != len(t_lst):
        #print("in len")
        return False
    n = len(s_lst)
    for i in range(n):
        for j in range(i + 1, n):
            if s_lst[i] == s_lst[j]:
                if t_lst[i] != t_lst[j]:
                    #print("in loop")
                    return False
    #print("in true")
    return True

# Print the Output
if are_isomorphic(s, t):
    print("True")
else:
    print("False")

```
