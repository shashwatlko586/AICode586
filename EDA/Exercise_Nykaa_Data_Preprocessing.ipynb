## Task 3: Handle missing values

### Subtask:
Identify columns with missing values, replace missing values in string columns with 'nodata', and replace missing numeric values with the median.


**Reasoning**:
The subtask requires identifying and handling missing values. First, I will display the missing value counts, then iterate through the columns to fill string columns with 'nodata' and numeric columns with their median, and finally verify the changes.



```python
print("Missing values before handling:")
display(df_nykaa.isnull().sum())

for column in df_nykaa.columns:
    if df_nykaa[column].isnull().any():
        if df_nykaa[column].dtype == 'object':
            df_nykaa[column].fillna('nodata', inplace=True)
        elif pd.api.types.is_numeric_dtype(df_nykaa[column]):
            median_value = df_nykaa[column].median()
            df_nykaa[column].fillna(median_value, inplace=True)

print("\nMissing values after handling:")
display(df_nykaa.isnull().sum())
```

**Reasoning**:
The previous code generated `FutureWarning` messages due to the use of `inplace=True` in `fillna`. To resolve this and ensure compatibility with future pandas versions, I will modify the code to assign the results of `fillna` back to the DataFrame columns instead of using the `inplace` argument.



```python
print("Missing values before handling:")
display(df_nykaa.isnull().sum())

for column in df_nykaa.columns:
    if df_nykaa[column].isnull().any():
        if df_nykaa[column].dtype == 'object':
            df_nykaa[column] = df_nykaa[column].fillna('nodata')
        elif pd.api.types.is_numeric_dtype(df_nykaa[column]):
            median_value = df_nykaa[column].median()
            df_nykaa[column] = df_nykaa[column].fillna(median_value)

print("\nMissing values after handling:")
display(df_nykaa.isnull().sum())
```

## Task 5: Standardise column names

### Subtask:
Clean column names by removing whitespaces, converting to lowercase, replacing spaces with underscores, ensuring no consecutive underscores, and removing leading/trailing underscores.


**Reasoning**:
To standardize the column names, I will apply a series of string cleaning operations using Python's `re` module to remove whitespaces, convert to lowercase, replace spaces with underscores, and handle consecutive or leading/trailing underscores. After cleaning, I will assign the new column names back to the DataFrame and display them for verification.



```python
import re

# Get current column names
current_columns = df_nykaa.columns

# List to store cleaned column names
cleaned_columns = []

# Iterate and clean each column name
for col in current_columns:
    # Remove leading and trailing whitespaces
    col = col.strip()
    # Convert to lowercase
    col = col.lower()
    # Replace spaces with underscores, handling multiple spaces
    col = re.sub(r'\s+', '_', col)
    # Remove any consecutive underscores
    col = re.sub(r'_+', '_', col)
    # Remove leading or trailing underscores
    col = col.strip('_')
    cleaned_columns.append(col)

# Assign the cleaned list of column names back to the DataFrame
df_nykaa.columns = cleaned_columns

print("Updated column names:")
display(df_nykaa.columns)
```

## Engineer new features

### Subtask:
Create 'price_category' based on product price and 'has_reviews' based on review count, then find the mode of 'price_category'.


**Reasoning**:
I will create the 'price_category' column using conditional logic and the 'has_reviews' column based on the 'product_reviews_count', then calculate and display the mode of 'price_category'.



```python
df_nykaa['price_category'] = df_nykaa['product_price'].apply(lambda x: 'low' if x < 500 else ('medium' if 500 <= x <= 2000 else 'high'))
df_nykaa['has_reviews'] = df_nykaa['product_reviews_count'].apply(lambda x: 'yes' if x >= 1 else 'no')

mode_price_category = df_nykaa['price_category'].mode()[0]
print(f"Mode of 'price_category': {mode_price_category}")

display(df_nykaa.head())
```

## Treat outliers

### Subtask:
Identify and treat outliers in the 'product_price' column using the IQR method by capping them to the upper or lower limits.


**Reasoning**:
To identify and treat outliers in the 'product_price' column, I will first calculate Q1, Q3, IQR, Upper Limit (UL), and Lower Limit (LL). Then, I will count the outliers and cap them at the calculated limits.



```python
Q1 = df_nykaa['product_price'].quantile(0.25)
Q3 = df_nykaa['product_price'].quantile(0.75)
IQR = Q3 - Q1

UL = Q3 + 1.5 * IQR
LL = Q1 - 1.5 * IQR

outliers_count = df_nykaa[(df_nykaa['product_price'] < LL) | (df_nykaa['product_price'] > UL)].shape[0]
print(f"Number of outliers identified: {outliers_count}")

# Cap the outliers
df_nykaa['product_price'] = df_nykaa['product_price'].clip(lower=LL, upper=UL)

print("\nDescriptive statistics for 'product_price' after outlier treatment:")
display(df_nykaa['product_price'].describe())
```
