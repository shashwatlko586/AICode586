# Task 1
## Inspect the data
- Load the data
- Study the shape of the data
- Check the data for data types and missing values
- Display top 5 rows

```python
print('Shape of the data:', df_nykaa.shape)
print('\nData information (data types and non-null counts):')
df_nykaa.info()
print('\nSummary statistics:')
display(df_nykaa.describe(include='all'))
```

# Task 2
## Analyse product price distribution
- Plot the discrete histogram of `'product_price'` with 5 bins
- Look for trends in the distribution

```python
plt.figure(figsize=(10, 6))
sns.histplot(df_nykaa['product_price'], bins=5, kde=False)
plt.title('Distribution of Product Price')
plt.xlabel('Product Price')
plt.ylabel('Count')
plt.show()
print("From the histogram, we can observe that most products are in the lower price ranges, with fewer products in the higher price brackets. The distribution appears right-skewed.")
```

# Task 3
## Analyse product brand counts
- Obtain the counts of all the brands in the dataset
- Find the top 5 brands based on the count in descending order
- Create a copy of the data frame with only these identified brands
- Check the shape of the data subset that you created

```python
brand_counts = df_nykaa['product_brand'].value_counts()
top_5_brands = brand_counts.head(5).index.tolist()
print('Top 5 Brands:', top_5_brands)
df_nykaa_subset = df_nykaa[df_nykaa['product_brand'].isin(top_5_brands)].copy()
print('Shape of the subset data:', df_nykaa_subset.shape)
```

**Note:** From this point onwards, please use the subset data that you created in the previous step

# Task 4
## Analyse price distribution by brand
- Obtain the mean product price of each brand

```python
mean_price_by_brand = df_nykaa_subset.groupby('product_brand')['product_price'].mean()
print('Mean Product Price by Brand:')
print(mean_price_by_brand)
```

# Task 5
## Analyse price distribution by product category
- Obtain the mean product price of each product category
- Sort the result in ascending order

```python
mean_price_by_category = df_nykaa_subset.groupby('product_category')['product_price'].mean().sort_values(ascending=True)
print('Mean Product Price by Category (sorted in ascending order):')
print(mean_price_by_category)
```

# Task 6
## Analyse product ratings by presence of reviews
- Obtain the mean product ratings for products that have at least one review and those that have no reviews

```python
mean_rating_by_reviews = df_nykaa_subset.groupby('has_reviews')['product_rating'].mean()
print('Mean Product Ratings by Presence of Reviews:')
print(mean_rating_by_reviews)
```

# Task 8
## Analyse relation between review counts and ratings
- Plot a scatter plot between `'product_reviews_count'` and `'product_rating'`
- Plot another scatter plot between natural logarithm of `'product_reviews_count'` and `'product_rating'`

```python
plt.figure(figsize=(12, 5))

# Scatter plot for product_reviews_count vs product_rating
plt.subplot(1, 2, 1)
sns.scatterplot(x='product_reviews_count', y='product_rating', data=df_nykaa_subset)
plt.title('Product Reviews Count vs Product Rating')
plt.xlabel('Product Reviews Count')
plt.ylabel('Product Rating')

# Scatter plot for natural logarithm of product_reviews_count vs product_rating
plt.subplot(1, 2, 2)
sns.scatterplot(x=np.log1p(df_nykaa_subset['product_reviews_count']), y='product_rating', data=df_nykaa_subset)
plt.title('Log(Product Reviews Count) vs Product Rating')
plt.xlabel('Log(Product Reviews Count)')
plt.ylabel('Product Rating')

plt.tight_layout()
plt.show()
```

# Task 9
## Analyse relation between product ratings and price
- Plot a scatter plot between `'product_rating'` and `'product_price'`
- Plot another scatter plot between `'product_rating'` and natural logarithm of `'product_price'`
- Plot a regression plot using `regplot` from `seaborn` between `'product_rating'` and natural logarithm of `'product_price'`

```python
plt.figure(figsize=(18, 5))

# Scatter plot for product_rating vs product_price
plt.subplot(1, 3, 1)
sns.scatterplot(x='product_rating', y='product_price', data=df_nykaa_subset)
plt.title('Product Rating vs Product Price')
plt.xlabel('Product Rating')
plt.ylabel('Product Price')

# Scatter plot for product_rating vs log(product_price)
plt.subplot(1, 3, 2)
sns.scatterplot(x='product_rating', y=np.log1p(df_nykaa_subset['product_price']), data=df_nykaa_subset)
plt.title('Product Rating vs Log(Product Price)')
plt.xlabel('Product Rating')
plt.ylabel('Log(Product Price)')

# Regression plot for product_rating vs log(product_price)
plt.subplot(1, 3, 3)
sns.regplot(x='product_rating', y=np.log1p(df_nykaa_subset['product_price']), data=df_nykaa_subset, scatter_kws={'alpha':0.6})
plt.title('Regression Plot: Product Rating vs Log(Product Price)')
plt.xlabel('Product Rating')
plt.ylabel('Log(Product Price)')

plt.tight_layout()
plt.show()
```

# Task 10
## Analyse relation between review counts, ratings, and brand
- Plot a scatter plot between natural logarithm of `'product_reviews_count'` and `'product_rating'` and set hue as `'product_brand'`
- Plot a regression plot using `regplot` from `seaborn` between natural logarithm of `'product_reviews_count'` and `'product_rating'` only for the brand `'LIME CRIME'`
- Plot another regression plot using `regplot` from `seaborn` between natural logarithm of `'product_reviews_count'` and `'product_rating'` only for the brand `'LAKME'`

```python
plt.figure(figsize=(18, 5))

# Scatter plot for log(product_reviews_count) vs product_rating with product_brand as hue
plt.subplot(1, 3, 1)
sns.scatterplot(x=np.log1p(df_nykaa_subset['product_reviews_count']), y='product_rating', hue='product_brand', data=df_nykaa_subset)
plt.title('Log(Reviews Count) vs Rating by Brand')
plt.xlabel('Log(Product Reviews Count)')
plt.ylabel('Product Rating')

# Regression plot for LIME CRIME
plt.subplot(1, 3, 2)
sns.regplot(x=np.log1p(df_nykaa_subset[df_nykaa_subset['product_brand'] == 'LIME CRIME']['product_reviews_count']),
            y=df_nykaa_subset[df_nykaa_subset['product_brand'] == 'LIME CRIME']['product_rating'],
            data=df_nykaa_subset, scatter_kws={'alpha':0.6})
plt.title('Regression Plot: LIME CRIME (Log(Reviews Count) vs Rating)')
plt.xlabel('Log(Product Reviews Count)')
plt.ylabel('Product Rating')

# Regression plot for LAKME
plt.subplot(1, 3, 3)
sns.regplot(x=np.log1p(df_nykaa_subset[df_nykaa_subset['product_brand'] == 'LAKME']['product_reviews_count']),
            y=df_nykaa_subset[df_nykaa_subset['product_brand'] == 'LAKME']['product_rating'],
            data=df_nykaa_subset, scatter_kws={'alpha':0.6})
plt.title('Regression Plot: LAKME (Log(Reviews Count) vs Rating)')
plt.xlabel('Log(Product Reviews Count)')
plt.ylabel('Product Rating')

plt.tight_layout()
plt.show()
```

# Task 11
## Analyse relation between ratings, price, and brand
- Plot a scatter plot between `'product_rating'` and `'product_price'` and set hue as `'product_brand'`
- Plot a scatter plot between `'product_rating'` and natural logarithm of `'product_price'` and set hue as `'product_brand'`
- Plot a regression plot using `regplot` from `seaborn` between `'product_rating'` and natural logarithm of `'product_price'` only for the brand `'HIMALAYA'`
- Plot a regression plot using `regplot` from `seaborn` between `'product_rating'` and natural logarithm of `'product_price'` only for the brand `'LIME CRIME'`

```python
plt.figure(figsize=(24, 6))

# Scatter plot for product_rating vs product_price with product_brand as hue
plt.subplot(1, 4, 1)
sns.scatterplot(x='product_rating', y='product_price', hue='product_brand', data=df_nykaa_subset)
plt.title('Product Rating vs Product Price by Brand')
plt.xlabel('Product Rating')
plt.ylabel('Product Price')

# Scatter plot for product_rating vs log(product_price) with product_brand as hue
plt.subplot(1, 4, 2)
sns.scatterplot(x='product_rating', y=np.log1p(df_nykaa_subset['product_price']), hue='product_brand', data=df_nykaa_subset)
plt.title('Product Rating vs Log(Product Price) by Brand')
plt.xlabel('Product Rating')
plt.ylabel('Log(Product Price)')

# Regression plot for HIMALAYA
plt.subplot(1, 4, 3)
sns.regplot(x='product_rating',
            y=np.log1p(df_nykaa_subset[df_nykaa_subset['product_brand'] == 'HIMALAYA']['product_price']),
            data=df_nykaa_subset[df_nykaa_subset['product_brand'] == 'HIMALAYA'], scatter_kws={'alpha':0.6})
plt.title('Regression Plot: HIMALAYA (Rating vs Log(Price))')
plt.xlabel('Product Rating')
plt.ylabel('Log(Product Price)')

# Regression plot for LIME CRIME
plt.subplot(1, 4, 4)
sns.regplot(x='product_rating',
            y=np.log1p(df_nykaa_subset[df_nykaa_subset['product_brand'] == 'LIME CRIME']['product_price']),
            data=df_nykaa_subset[df_nykaa_subset['product_brand'] == 'LIME CRIME'], scatter_kws={'alpha':0.6})
plt.title('Regression Plot: LIME CRIME (Rating vs Log(Price))')
plt.xlabel('Product Rating')
plt.ylabel('Log(Product Price)')

plt.tight_layout()
plt.show()
```
