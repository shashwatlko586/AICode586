```python
class Animal:
    def speak(self):
        print("Some sound")

class Dog(Animal):
    def speak(self):
        print("Bark")

class Cat(Animal):
    def speak(self):
        print("Meow")

animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```

```python
class Developer:
    def role(self):
        print("Writes code")

class Designer:
    def role(self):
        print("Designs UI")

# Polymorphic behaviour
def show_role(person):
    person.role()

d1 = Developer()
d2 = Designer()

show_role(d1)  # Output: Writes code
show_role(d2)  # Output: Designs UI
```
