```python
unique_brands_list = df_nykaa['Product Brand'].unique().tolist()
df_unique_brands = pd.DataFrame(unique_brands_list, columns=['Product Brand'])

specific_brands = ['Biotique', 'Himalaya', 'Lotus Herbals']
filtered_df_unique_brands = df_unique_brands[df_unique_brands['Product Brand'].isin(specific_brands)]

print(f"Unique Product Brands as DataFrame for {', '.join(specific_brands)}:")
display(filtered_df_unique_brands)
```

```python
brand_prices = df_nykaa.groupby('Product Brand')['Product Price'].unique().apply(list).reset_index()
brand_prices.rename(columns={'Product Price': 'Unique Prices'}, inplace=True)

# Filter for the specified brands
specific_brands = ['Biotique', 'Himalaya', 'Lotus Herbals']
filtered_brand_prices = brand_prices[brand_prices['Product Brand'].isin(specific_brands)]

print(f"Unique Product Brands and their Prices for {', '.join(specific_brands)}:")
display(filtered_brand_prices)
```

```python
specific_brands = ['Biotique', 'Himalaya', 'Lotus Herbals']
filtered_df_nykaa = df_nykaa[df_nykaa['Product Brand'].isin(specific_brands)]

# Get the price data for the specified brands
prices = filtered_df_nykaa['Product Price'].values

# Calculate bin edges for 10 bins
hist, bin_edges = np.histogram(prices, bins=10)

print("Bin Edges:")
for i in range(len(bin_edges) - 1):
    print(f"Bin {i+1}: {bin_edges[i]:.2f} - {bin_edges[i+1]:.2f}")

# Check counts per bin for each brand
print("\nCounts per brand per bin:")
for brand in specific_brands:
    brand_prices = filtered_df_nykaa[filtered_df_nykaa['Product Brand'] == brand]['Product Price'].values
    counts, _ = np.histogram(brand_prices, bins=bin_edges) # Use same bin edges
    print(f"Brand: {brand}")
    for i, count in enumerate(counts):
        print(f"  Bin {i+1} ({bin_edges[i]:.2f} - {bin_edges[i+1]:.2f}): {count} items")

# Identify bins with zero total count
zero_count_bins = []
for i in range(len(bin_edges) - 1):
    total_count_in_bin = 0
    for brand in specific_brands:
        brand_prices = filtered_df_nykaa[filtered_df_nykaa['Product Brand'] == brand]['Product Price'].values
        counts, _ = np.histogram(brand_prices, bins=bin_edges)
        total_count_in_bin += counts[i]
    if total_count_in_bin == 0:
        zero_count_bins.append(f"Bin {i+1} ({bin_edges[i]:.2f} - {bin_edges[i+1]:.2f})")

if zero_count_bins:
    print(f"\nBins with zero count values across all specified brands: {', '.join(zero_count_bins)}")
else:
    print("\nNo bins with zero count values across all specified brands.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_nykaa = pd.read_csv('nykaa.csv')

specific_brands = ['Biotique', 'Himalaya', 'Lotus Herbals']
filtered_df_nykaa = df_nykaa[df_nykaa['Product Brand'].isin(specific_brands)]

plt.figure(figsize=(10, 6))
sns.countplot(data=filtered_df_nykaa, x='Product Brand', hue='Product Brand', palette='viridis', legend=False)
plt.title('Count of Products for Biotique, Himalaya, and Lotus Herbals')
plt.xlabel('Product Brand')
plt.ylabel('Number of Products')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
specific_brands = ['Biotique', 'Himalaya', 'Lotus Herbals']
filtered_df_nykaa = df_nykaa[df_nykaa['Product Brand'].isin(specific_brands)]

# Calculate mean prices
mean_prices = filtered_df_nykaa.groupby('Product Brand')['Product Price'].mean().reset_index()

# Plot mean prices
plt.figure(figsize=(10, 6))
sns.barplot(data=mean_prices, x='Product Brand', y='Product Price', palette='viridis')
plt.title('Mean Product Prices for Biotique, Himalaya, and Lotus Herbals')
plt.xlabel('Product Brand')
plt.ylabel('Mean Product Price')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Calculate median prices
median_prices = filtered_df_nykaa.groupby('Product Brand')['Product Price'].median().reset_index()

# Plot median prices
plt.figure(figsize=(10, 6))
sns.barplot(data=median_prices, x='Product Brand', y='Product Price', palette='magma')
plt.title('Median Product Prices for Biotique, Himalaya, and Lotus Herbals')
plt.xlabel('Product Brand')
plt.ylabel('Median Product Price')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Calculate value counts for 'Product Brand'
brand_counts = df_nykaa['Product Brand'].value_counts()

# Get the top ten most represented companies
x = brand_counts.values[:10]
labels = brand_counts.index[:10]

# Plotting the pie chart
plt.figure(figsize=(10, 8))
plt.pie(x, labels=labels, autopct='%1.2f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Representation of Top 10 Most Represented Product Brands')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
# Select the relevant columns
correlation_data = df_nykaa[['Product Price', 'Product Rating', 'Product Reviews Count']]

# Calculate the correlation matrix
correlation_matrix = correlation_data.corr()

# Plot the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', linewidths=.5)
plt.title('Correlation Heatmap of Product Price, Rating, and Reviews Count')
plt.show()
```

```python
# Select the relevant numerical columns for the pairplot
pairplot_data = df_nykaa[['Product Price', 'Product Rating', 'Product Reviews Count']]

# Create the pairplot
sns.pairplot(pairplot_data)
plt.suptitle('Pairplot of Product Price, Rating, and Reviews Count', y=1.02) # Adjust title position
plt.show()
```
