```python
df_swiggy = pd.read_csv('swiggy.csv')
df_swiggy.head()
```

What is the mean price across all restaurants in Bangalore? Round your answer to two decimal places.

```python
bangalore_restaurants = df_swiggy[df_swiggy['City'] == 'Bangalore']
mean_price_bangalore = bangalore_restaurants['Price'].mean()
print(f"Mean price in Bangalore: {mean_price_bangalore:.2f}")
```

```python
bangalore_restaurants_df = df_swiggy[df_swiggy['City'] == 'Bangalore']
total_ratings_bangalore = bangalore_restaurants_df['Total ratings'].sum()
print(f"Total number of ratings across all restaurants in Bangalore: {total_ratings_bangalore}")
```
