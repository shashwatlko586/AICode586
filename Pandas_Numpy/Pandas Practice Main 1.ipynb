# **Generate a barplot showing the median number of Google review ratings in lakhs for each zone, broken down by whether a DSLR is allowed or not. Turn off error bars.**

# **Select the zone from the options below which has a higher median number of ratings at tourist spots where DSLRs are allowed than in those where they are not.**

```python
plt.figure(figsize = (5, 3))
sns.barplot(
    data = tourist_spots,
    x = 'Zone',
    y = 'Number of google review in lakhs',
    hue = 'DSLR Allowed',
    estimator = 'median',
    errorbar = None,
)
plt.xlabel('Zones')
plt.xticks(rotation = 20)
plt.ylabel('Google review ratings in lakhs')
plt.title('Median number of Google review ratings in lakhs for each zone with DSLR permissions')

plt.show()
```
