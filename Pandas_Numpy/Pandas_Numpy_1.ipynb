```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
b = [[5, 6], [7, 8]]
print(np.append(a, b, axis=1))
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = np.array([[7, 8], [9, 10]])
d = np.array([[11, 12], [13, 14]])

print("np.concatenate((a, b), axis = 0)")
print(np.concatenate((a, b), axis = 0))

print("np.concatenate((a2, b2), axis = 1)")
a2 = a.reshape(1, -1)
b2 = b.reshape(1, -1)
print(a)
print(a2)
print(np.concatenate((a2, b2), axis = 1))

print("np.concatenate((a3, b3), axis = 1)")
a3 = a.reshape(-1, 1)
b3 = b.reshape(-1, 1)
print(a)
print(a3)
print(np.concatenate((a3, b3), axis = 1))
```
