# **Compute the total population of a state**

# **Instructions**

Problem Statement

You are given district-wise population data, and you must write Python code using Pandas to compute the total population of a state.

Write Python code using Pandas to calculate the total population of a state.



Data Description

The dataset is a CSV file containing district-wise population details. The shape of the dataset is (640, 119). Each row represents one district and includes the district name, the state to which it belongs, and its population. The relevant columns in the dataset are State name(string), District (string), and Population (integer). All entries in the dataset are valid and contain no missing values. The dataset uses default integer row indices.



Input Format

A single string representing the state name

Output Format

A single integer representing the total population of the state



Sample Case 1 Input

ASSAM

# **Sample Case 1 Output**

31205576



Sample Case 2 Input

TAMIL NADU

Sample Case 2 Output

72147030



Constraints

Case changes in state names are not accounted for
Any input string other than a state name within the dataset (Indian states) are not accounted for
The integer columns contain only positive values

```python
# Do not edit this data and input processing section
import pandas as pd
df = pd.read_csv('https://d3ejq4mxgimsmf.cloudfront.net/india_census_2011-ab83d5de80f640099aaa4463a76de43d.csv')
state = input().strip()

# Write your code here
sum_ser = df.groupby('State name')['Population'].sum()

# Do not edit this output processing section
print(sum_ser[state])
```

```python
# Do not edit this data read section
import pandas as pd
df = pd.read_csv('https://d3ejq4mxgimsmf.cloudfront.net/india_census_2011-ce94dcde69614e2686012fcff798c587.csv')

# Write your code here
df['Literate_rate'] = df['Literate'] / df['Population']

# Find the index of the row with the maximum literacy rate for each state
idx_max_literacy_per_state = df.groupby('State name')['Literate_rate'].idxmax()

# Get the district names corresponding to these indices. This will still be indexed by the original df integer index.
temp_district_names = df.loc[idx_max_literacy_per_state, 'District name']

# Create the final ser_out with 'State name' as index and district names as values
ser_out = pd.Series(data=temp_district_names.values, index=idx_max_literacy_per_state.index)

# Do not edit this input-output processing section
print(ser_out[input().strip()])
```
