## Target Variable Definition

### Subtask:
Set 'Grad.Rate' as the target variable and all other columns as predictors, as specified in Task 3.


**Reasoning**:
As instructed, I will create the target variable `y` by selecting the 'Grad.Rate' column and the predictor variables `X` by selecting all other columns from the `college` DataFrame.



```python
y = college['Grad.Rate']
X = college.drop('Grad.Rate', axis=1)
```

## Data Preprocessing

### Subtask:
Convert categorical variables into numeric format using one-hot encoding with `pd.get_dummies()` and `drop_first = True` to avoid multicollinearity, as per Task 4.


**Reasoning**:
First, I'll identify the categorical columns in the `X` DataFrame using `select_dtypes(include='object')`. Then, I will apply one-hot encoding to these columns using `pd.get_dummies()` with `drop_first=True` to convert them into a numeric format while avoiding multicollinearity, as instructed.



```python
categorical_cols = X.select_dtypes(include='object').columns
X = pd.get_dummies(X, columns=categorical_cols, drop_first=True)

print("DataFrame X after one-hot encoding:")
print(X.head())
```

## Train-Validation Split

### Subtask:
Split the data into training (70%) and validation (30%) sets using `train_test_split()` with `random_state = 42` for reproducibility, as per Task 5.


**Reasoning**:
I will split the feature and target data into training and validation sets using `train_test_split` with the specified parameters (`test_size=0.3`, `random_state=42`) and assign them to the correct variables.



```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

# Task 6
### Model Training
- Fit a linear regression model on the training data using `LinearRegression()` from `sklearn`

# Task 6
### Model Training
- Fit a linear regression model on the training data using `LinearRegression()` from `sklearn`

**Reasoning**:
The previous step introduced the 'Model Training' task. Now, I need to implement the model training by initializing and fitting a `LinearRegression` model using the `X_train` and `y_train` data.



```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Linear Regression model fitted successfully.")
```

# Task 7
### Model Parameters
- Display model coefficients and intercept values using the attributes `.coef_` and `.intercept_`

# Task 7
### Model Parameters
- Display model coefficients and intercept values using the attributes `.coef_` and `.intercept_`

**Reasoning**:
Now that Task 7 has been introduced, I will generate the Python code to display the trained model's coefficients and intercept using `model.coef_` and `model.intercept_` as instructed.



```python
print("Model Coefficients:", model.coef_)
print("Model Intercept:", model.intercept_)
```

# Task 8
### Predictions
- Generate predictions on both training and testing sets

# Task 8
### Predictions
- Generate predictions on both training and testing sets

**Reasoning**:
Now that Task 8 has been introduced, I will generate predictions for both the training and testing datasets using the trained linear regression model's `predict()` method.



```python
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

print("Predictions generated for both training and testing sets.")
```

# Task 9
### Model Performance
- Evaluate model performance using R² and MAE on both training and testing sets

**Reasoning**:
Now that Task 9 has been introduced, I will calculate the R² and Mean Absolute Error (MAE) for both the training and testing sets using the generated predictions and actual target values, as instructed.



```python
r2_train = r2_score(y_train, y_train_pred)
mae_train = mean_absolute_error(y_train, y_train_pred)

r2_test = r2_score(y_test, y_test_pred)
mae_test = mean_absolute_error(y_test, y_test_pred)

print(f"Training R\u00b2 Score: {r2_train:.2f}")
print(f"Training MAE: {mae_train:.2f}")
print(f"Testing R\u00b2 Score: {r2_test:.2f}")
print(f"Testing MAE: {mae_test:.2f}")
```
