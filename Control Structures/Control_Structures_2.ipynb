# **Find Greatest Common Divisor of Two Numbers**

# **Instructions**

Write a program to find  Greatest Common Divisor (GCD) of Two Positive Numbers.



The Greatest Common Divisor (GCD) of two positive integers is the largest positive number that divides both numbers exactly without leaving a remainder.




For example, the GCD of 48 and 18 is 6 because 6 is the biggest number that divides both 48 and 18 completely.



Input Format: Two integers given on separate lines.

Output Format: Single integer representing the GCD of the two numbers.




Sample Input:

48

18




# **Sample Output:**

6

```python
# Taking input
num1 = int(input())
num2 = int(input())

def gcd(a, b):
    # Write your code here
    for i in range( b if a >= b else a , 0 , -1 ):
        if a % i == 0 and b % i == 0: return i

# Print the output
print(gcd(num1, num2))
```

Check If a Given Number is Prime or Not.

Instructions

Write a program that checks if a given number is prime or not.

A prime number is a positive integer greater than 1 that has no divisors other than 1 and itself.



Constraints: The input number should be a positive integer greater than 1.




 Input Format:

Single integer n.




Output Format:

"Prime" if the number is prime, otherwise "Not Prime".



Sample Input:

7

Sample Output:

Prime

```python
# Taking input
num = int(input())

def is_prime(n):
    # Write your code
    if n <= 1:
        return "Not Prime"

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return "Not Prime"

    return "Prime"


# Print the output
print(is_prime(num))
```

 Input Format:

# **Reverse a Number Without Converting to String**

# **Instructions**

Write a program to reverse a Non-Negative Integer without converting it to a String.

The program should take an Integer and return the Integer Reversed.



Input Format:

Non-Negative Integer.



Output Format:

The Reversed Integer.



Constraints:

The input is a Non-Negative Intege.



Sample Input:

12345

# **Sample Output:**

54321

```python
# Taking input
num = int(input())

def reverse_number(n):
    # Write your code
    i = 0
    while n != 0 :
        i = i * 10 + n % 10
        n = n // 10
    return i
# Print the output
print(reverse_number(num))
```

The sum of the squares: 1² + 2² + ... + n²

Return the difference: (square of the sum) - (sum of the squares).

An integer representing the difference between the square of the sum and the sum of the squares of the first n natural numbers.

# **Sum of Digits Until Single Digit**

# **Instructions**

Write a Python function 'sum_until_single_digit(num)' that takes a Non-Negative Integer number and repeatedly sums its digits until the result is a single digit. Return that single-digit sum.



For example, given number = 9875,
The sum of digits is 9+8+7+5 = 29,

then sum digits of 29 → 2+9 = 11, and

Finally, sum the digits of 11 → 1+1 = 2.

The function returns 2.



Input Format:

A single non-negative integer num.



Output Format:

Return the single-digit integer obtained by repeatedly summing the digits of the number.



Constraints:

0 <= n<= 109



Sample Input:

9875

# **Sample Output:**

2

```python
# Taking input
num = int(input())

def sum_until_single_digit(num):
    # Write your code
    sum_until_single_digit_value = 0
    while num != 0 :
        sum_until_single_digit_value += num % 10
        if num // 10 == 0 :
          num = sum_until_single_digit_value
          sum_until_single_digit_value = 0
        else : num = num // 10
        # print ('sum_until_single_digit_value : ' + str(sum_until_single_digit_value))
        # print ('num : ' + str(num))
        if sum_until_single_digit_value == 0 and num < 10 :
          break
    return num

# Print the output
print(sum_until_single_digit(num))
```

# **Print Zigzag Number Pattern**

# **Instructions**

Write a Python function print_zigzag(n) that prints numbers from 1 to n in a zigzag pattern across three rows.

The zigzag pattern follows this repeating cycle every 4 numbers:

1st number in row 1
2nd number in row 2
3rd number in row 3
4th number in row 2


Each number is followed by a space.

Positions without a number should be filled with two spaces "  " to maintain alignment.



Input Format:

A single integer n representing the total number to print in the zigzag pattern.



Output Format:

Print 3 lines, each containing the numbers arranged in the zigzag pattern separated by spaces, with appropriate spacing for alignment.



Constraints:

1 <= n <=11



Sample Input:

10



# **Sample Output:**

1       5       9    

  2   4   6   8   10    

     3       7      

```python
# Taking input
num = int(input())

def print_zigzag(n):
    row1 = ""
    row2 = ""
    row3 = ""
    # Write your code here
    for i in range(1, n + 1):
        pos = (i - 1) % 4

        if pos == 0:          # Row 1
            row1 += str(i) + "   "
            row2 += "    "
            row3 += "    "
        elif pos == 1:        # Row 2
            row1 += "    "
            row2 += str(i) + "   "
            row3 += "    "
        elif pos == 2:        # Row 3
            row1 += "    "
            row2 += "    "
            row3 += str(i) + "   "
        else:                 # Row 2 again
            row1 += "    "
            row2 += str(i) + "   "
            row3 += "    "

    print(row1)
    print(row2)
    print(row3)
# Print the output
print_zigzag(num)
```

# **Check if a Number is a Strong Number**

# **Instructions**

Write a program to check whether a given number is a Strong Number or not.



A Strong Number is a number whose sum of the factorial of each digit is equal to the number itself.



For example,

145 is a strong number because: 1! + 4! + 5! = 1 + 24 + 120 = 145.



Input Format:

A single integer number.



Output Format:

Print True if the number is a Strong Number; otherwise, print False.



Sample Input:

145



# **Sample Output:**

True

```python
# Taking input
n=int(input())

def factorial(n):
   # Write Your Code here
   if n == 0 or n == 1:
        return 1
   return (n * factorial(n - 1))

def is_strong_number(number):
   # Write Your Code here
   sum_digit = 0
   for ch in str(number) :
      sum_digit += factorial(int(ch))
   return ( 'True' if sum_digit == number else 'False')


# Print the output
print(is_strong_number(n))
```

Write a program to calculate the price of a movie ticket based on the customer's age and the showtime.



The Pricing Rules are:

Children (age < 12):
Ticket costs 5 regardless of showtime.
Seniors (age >= 60):
Ticket costs 7 regardless of showtime.
Adults (age 12 to 59):
Matinee shows (before 5 PM):10

Constraints:

Age must be a non-negative integer.
The show hour must be an integer between 0 and 23 inclusive.
Children are defined as those under the age of 12.
Seniors are defined as those aged 60 or older.

Output Format:

An integer representing the ticket price in dollars.



Sample Input:

10

14

# **Simple ATM Cash Withdrawal Simulation**

# **Instructions**

Write a program to simulate an ATM cash withdrawal.



The program should take the withdrawal amount as input and verify its validity based on the available ATM balance and withdrawal rules.

The withdrawal amount must be a multiple of 100.
The withdrawal amount must not exceed the available balance.
The withdrawal amount must be greater than zero.


If all conditions are met, deduct the amount from the balance and print the remaining balance. Otherwise, print the appropriate error message.



Input Format:

A single integer representing the withdrawal amount. Decimal or floating-point numbers are not accepted.



Output Format:

If successful, print: "Transaction successful. Remaining balance: <balance>".
If the amount is not a multiple of 100, print: "Error: Amount must be a multiple of 100".
If the amount exceeds the balance, print: "Error: Insufficient funds".
If the amount is less than or equal to zero, print: "Error: Invalid withdrawal amount".


Sample Input:

500



# **Sample Output:**

Transaction successful. Remaining balance: 1500

```python
# Available balance in the ATM
balance = 2000

# Taking input
withdrawal_amount = int(input())

def atm_withdraw(amount, balance):
    # Write your code here
    if amount <= 0:
        return ("Error: Invalid withdrawal amount")
    elif amount % 100 != 0:
        return ("Error: Amount must be a multiple of 100")
    elif amount > balance:
        return ("Error: Insufficient funds")
    else:
        balance -= amount
        return (f"Transaction successful. Remaining balance: {balance}")

# Print the output
print(atm_withdraw(withdrawal_amount, balance))
```
