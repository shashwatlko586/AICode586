# **Check Armstrong Number**

# **Instructions**

 Write a function that checks whether a number is an Armstrong number. A number is Armstrong if the sum of the cubes of its digits is equal to the number itself.

***Here inputs will be provided only for 3 digit numbers not otherwise***


Input Format:

An integer num



Output Format:

Yes if the number is Armstrong, else No



Constraints:

0 <= num <= 9999



Sample Input:

153



# **Sample Output:**

Yes

```python
#Taking input
# Here inputs will be provided only for 3 digit numbers not otherwise
num = int(input())

def is_armstrong(num):
    # Write your code
    def sum_of_cubes_of_digit(num):
        if num > 0 and num < 10000 :
            return ( ( ( num % 10 ) ** 3 ) + sum_of_cubes_of_digit( num // 10 ))
        else :
            return 0
    if num >= 0 and num < 10000 :
        return ("Yes" if num == sum_of_cubes_of_digit(num) else "No")
    else :
        return "Out of constraints Scope"

# Print the output
print(is_armstrong(num))
```

# **Recursive Digit Product**

# **Instructions**

Write a Recursive Function 'digit_product(n)' that returns the product of all digits in the number n.



Input Format:

An integer n



Output Format:

An integer representing the product of digits



Constraints:

0 <= n <= 100000
Assume n will never contain a 0 digit for now




Sample Input:

234



# **Sample Output:**

24

```python
# Taking input
n = int(input())

def digit_product(n):
    # Write your code
    if  n == 0 :
        return 0
        # Assume n will never contain a 0 digit for now
    elif n > 0 and n <= 100000 :
        return (n % 10 * ( digit_product( n // 10 ) if ( n // 10 != 0 ) else 1 ) )
    else :
        return "Out of constraints Scope"

# Print the output
print(digit_product(n))
```

# **Basic Calculator with User-Defined Functions**

# **Instructions**

Write a function 'calculator(x, y, op)' that takes two numbers and an operator ('+', '-', '*', '/') and returns the result of applying the operator. Assume valid input.



Input Format:

Three inputs: two integers x, y and a character op



Output Format:

A float or integer representing the result



Constraints:

-100 <= a, b <= 100
For division, assume b != 0




Sample Input:

4  2  *



# **Sample Output:**

8

```python
# Taking input
x = int(input())
y = int(input())
op = input()

def calculator(x, y, op):
    # Write your code
    match op:
        case '+':
            return (lambda a, b: a + b)(x, y)
        case '-':
            return (lambda a, b: a - b)(x, y)
        case '*':
            return (lambda a, b: a * b)(x, y)
        case '/':
            # Added a check for division by zero
            return (lambda a, b: a / b)(x, y) if y != 0 else "Error: Division by zero"
        case _:
            return "Invalid Operator"

# Print the output
print(calculator(x, y, op))
```
